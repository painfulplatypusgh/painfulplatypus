$$\hat{w} = arg min\frac{1}{2}∑^{N}_{i=1}a^{(i)}(y^{(i)}-w^{T}x^{(i)})^2+\frac{λ}{2}||w||^2$$

=$$\frac{1}{2}(y-Xw)^TA(y-Xw)+\frac{\lambda}{2}w^Tw$$

=
$$\frac{1}{2}(y^TAy-y^TAXw-w^TX^TAy + w^TX^TAXw) +\frac{\lambda}{2}w^Tw$$

Since A is symmetric: $$y^TAXw = w^TX^TAy$$

So:
$$J(w) = \frac{1}{2}y^TAy - w^TX^TAy + \frac{1}{2}w^TX^TAXw + \frac{\lambda}{2}w^Tw$$
=
Gradient with respect to w:
$$∇_wJ(w) = -X^TAy+X^TAXw +λw$$

Now solve:
$$∇_wJ(w) = 0$$
$$-X^TAy + X^TAXw + \lambda Iw = 0$$
$$(X^TAX + \lambda I)w = X^TAy$$
So:
$$\hat{w} = (X^TAX+\lambda I)^{-1}X^TAy$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import seaborn as sns
import torch.optim as optim
import statsmodels.api as sm
```

```python
class CauchyLoss(nn.Module):
    def __init__(self, c=1.0):
        super().__init__()
        self.c = c

    def forward(self, y_pred, y_true):
      residual = y_true - y_pred
      loss = (self.c**2/2.0) * torch.log(1.0+(residual/self.c)**2)
      return torch.mean(loss)

class CauchyRegression:
  def __init__(self, input_dim=4, c=1.0, lr=0.01, epochs=2000):
    self.c = c
    self.lr = lr
    self.epochs = epochs

    self.model = nn.Linear(input_dim, 1)
    self.criterion = CauchyLoss(c=self.c)
    self.optimizer = optim.Adam(self.model.parameters(), lr=self.lr)
    self.X_mean = None
    self.X_std = None

  def fit(self, X, y):
    X_arr = np.array(X, dtype=np.float32)
    y_arr = np.array(y, dtype=np.float32).reshape(-1,1)

    self.X_mean = np.mean(X_arr, axis=0)
    self.X_std = np.std(X_arr, axis=0)

    with torch.no_grad():
      self.model.bias.fill_(float(np.mean(y_arr)))

    X_scaled = (X_arr - self.X_mean) / self.X_std
    X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
    y_tensor = torch.tensor(y_arr, dtype=torch.float32)

    for epoch in range(self.epochs):
      self.optimizer.zero_grad()
      predictions = self.model(X_tensor)
      loss = self.criterion(predictions, y_tensor)
      loss.backward()
      self.optimizer.step()

    self.final_loss = loss.item()

  def predict(self, X):
    X_arr = np.array(X, dtype=np.float32)
    X_scaled = (X_arr - self.X_mean) / self.X_std
    X_tensor = torch.tensor(X_scaled, dtype=torch.float32)
    with torch.no_grad():
      predictions = self.model(X_tensor)
    return predictions.numpy().flatten()

  def get_unscaled_coefficients(self):
    weights = self.model.weight.detach().numpy().flatten()
    bias = self.model.bias.detach().item()

    unscaled_weights = weights/self.X_std
    unscaled_intercept = bias - np.sum(weights * self.X_mean/self.X_std)

    return unscaled_intercept, unscaled_weights

url = "https://raw.githubusercontent.com/rahulbhadani/Datasets/refs/heads/master/CombinedCyclePowerPlant/Folds5x2_pp.csv"
df = pd.read_csv(url)

#Q1: Scatterplot Matrix
sns.pairplot(df, vars=['AT', 'V', 'AP', 'RH', 'PE'], diag_kind='kde', plot_kws={'alpha':0.3, 's':10})
plt.suptitle("Scatterplot Matrix of Features and Target")
plt.show()

#Q2: Fit Cauchy Regression
X = df[['AT', 'V', 'AP', 'RH']]
y = df['PE']
cauchy_model = CauchyRegression(input_dim=4, c=1.0, lr=0.01, epochs=3000)
cauchy_model.fit(X, y)
y_pred = cauchy_model.predict(X)
residuals = y_pred - y.values
#Q3: Coefficient Values
w0, (w1, w2, w3, w4) = cauchy_model.get_unscaled_coefficients()
print(f"Coefficient Values: w0: {w0}, w1: {w1}, w2: {w2}, w3: {w3}, w4: {w4}")

#Q4: Final Loss Value
print(f"Final Loss Value: {cauchy_model.final_loss}")

#Q5: 95% CI
X_sm = sm.add_constant(X)
ols_model = sm.OLS(y, X_sm).fit()
conf_intervals = ols_model.conf_int(alpha=0.05)
print(f"95% Confidence Intervals: {conf_intervals}")

#Q6: Residual Plots
fig, axes = plt.subplots(2, 2)
features = ['AT', 'V', 'AP', 'RH']

for idx, col in enumerate(features):
  ax = axes.flat[idx]
  ax.scatter(df[col], residuals, alpha=0.3, s=10)
  ax.set_xlabel(col)
  ax.set_ylabel('Residual')
  ax.set_title(f"Residual Plot vs {col}")

  plt.tight_layout()
plt.show()




```

```python
from sklearn.datasets import fetch_openml
from sklearn.metrics import(accuracy_score, classification_report, confusion_matrix, f1_score, precision_score, recall_score, roc_auc_score)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

#Load dataset "Adult Census Income" (OpenML ID: 1590)
#Samples: 48842
#Features: 14
url = "https://raw.githubusercontent.com/pooja2512/Adult-Census-Income/refs/heads/master/adult.csv"
df = pd.read_csv(url)

df["income"] = df["income"].astype(str).str.strip()
y = (df["income"] == ">50K").astype(int)

selected_features = ['age', 'education.num', 'capital.gain', 'hours.per.week']
X = df[selected_features].copy()

X = X.replace("?", np.nan).dropna()
y = y.loc[X.index]

X_np = X.values.astype(np.float32)
y_np = y.values.astype(np.float32)

X_train, X_test, y_train, y_test = train_test_split(X_np, y_np, test_size=0.2, random_state=42, stratify=y_np)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train, dtype=torch.float32).unsqueeze(1)
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32).unsqueeze(1)

#Class Inbalance Check

unique, counts = np.unique(y_np, return_counts=True)
class_dist = dict(zip(unique, counts))
if max(counts) / min(counts) > 1.5:
  print("Class imbalance detected")

#Implement Logistic Regression
class LogisticRegression(nn.Module):
  def __init__(self, input_dim):
    super().__init__()
    self.linear = nn.Linear(input_dim, 1)

  def forward(self, x):
    return torch.sigmoid(self.linear(x))

input_dim = X_train_tensor.shape[1]
model = LogisticRegression(input_dim)
criterion = nn.BCELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

#Train model
epochs = 50
batch_size = 64
n_samples = X_train_tensor.shape[0]

for epoch in range(epochs):
  permutation = torch.randperm(n_samples)
  epoch_loss = 0.0
  for i in range(0, n_samples, batch_size):
    indices = permutation[i:i+batch_size]
    batch_x, batch_y = X_train_tensor[indices], y_train_tensor[indices]

    outputs = model(batch_x)
    loss = criterion(outputs, batch_y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    epoch_loss += loss.item() * len(batch_x)

  if(epoch + 1) % 10 == 0 or epoch == 0:
    avg_loss = epoch_loss / n_samples
    print(f"Epoch [{epoch+1}/{epochs}] - Loss: {avg_loss:.4f}")

#Evaluation on test dataset
model.eval()
with torch.no_grad():
  y_pred_probs = model(X_test_tensor).numpy()
  y_pred = (y_pred_probs >= 0.5).astype(int)

acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_pred_probs)
cm = confusion_matrix(y_test, y_pred)

print("\nEvaluation Metrics")
print(f"Accuracy: {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall: {rec:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"ROC AUC Score: {roc_auc:.4f}")
print("\nConfusion Matrix:")
print(cm)
```
