7.

Dropout is a technique where, during each training iteration, a random subset of neurons are temporarily deactivated (outputs set to zero). This can prevent overfitting by preventing the network from memorizing specific training examples. It also eliminates co-adaptation of neurons, which is where neurons in a layer can become overly dependent on neighboring neurons to correct their mistakes. Since any neighbor might be deactivated at any step, every neuron is forced to learn independently useful features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

#Load dataset
url = "https://raw.githubusercontent.com/rahulbhadani/Datasets/refs/heads/master/Diabetes/diabetes.csv"
df = pd.read_csv(url)

#Separate features
X = df.drop(columns=['Outcome'])
y = df['Outcome']

#Standardize before PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

#Part 1: PCA with two components
pca_2 = PCA(n_components=2)
X_pca2 = pca_2.fit_transform(X_scaled)
var_2 = pca_2.explained_variance_ratio_
print("---PCA with 2 components---")
print(f"PC1 explains {var_2[0]*100:.2f}% of the variance")
print(f"PC2 explains {var_2[1]*100:.2f}% of the variance")

#Part 2: 2d plot analysis
plt.figure(figsize=(8,6))
for outcome, color, label in [(0, 'steelblue', 'Outcome 0 (No Diabetes)'), (1, 'darkorange', 'Outcome 1 (Diabetes)')]:
    mask = (y == outcome)
    plt.scatter(X_pca2[mask, 0], X_pca2[mask, 1], c=color, label=label)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.title('2d PCA Analysis')
plt.show()
print("The two classes are not clearly distinguishable.")

#Part 3: PCA with 3 components
pca_3 = PCA(n_components=3)
X_pca3 = pca_3.fit_transform(X_scaled)
var_3 = pca_3.explained_variance_ratio_
print("---PCA with 3 components---")
print(f"PC1 explains {var_3[0]*100:.2f}% of the variance")
print(f"PC2 explains {var_3[1]*100:.2f}% of the variance")
print(f"PC3 explains {var_3[2]*100:.2f}% of the variance")

#Part 4: 3d plot analysis
fig = plt.figure(figsize=(9,7))
ax = fig.add_subplot(111, projection='3d')

for outcome, color, label in [(0, 'steelblue', 'Outcome 0 (No Diabetes)'), (1, 'darkorange', 'Outcome 1 (Diabetes)')]:
    mask = (y == outcome)
    ax.scatter(X_pca3[mask, 0], X_pca3[mask, 1], X_pca3[mask, 2], c=color, label=label)

ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.set_zlabel('PC3')
ax.set_title('3d PCA Analysis')
plt.legend
plt.show()
print("The two classes are slightly more distinguishable than in the 2d plot, but they still exhibit heavy overlap.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Generate a dataset
X, y = make_circles(n_samples=400, noise=0.1, factor=0.3, random_state=1)

# Helper function to plot decision boundaries
def plot_decision_boundary(model, X, y, ax, title):
    h = 0.02
    x_min, x_max = X[:, 0].min() - 0.3, X[:, 0].max() + 0.3
    y_min, y_max = X[:, 1].min() - 0.3, X[:, 1].max() + 0.3
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    ax.contourf(xx, yy, Z, cmap=plt.cm.coolwarm, alpha=0.35)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.coolwarm, edgecolors='k', s=30)
    ax.set_title(title)
    ax.set_xlabel('$x_1$')
    ax.set_ylabel('$x_2$')

plt.figure(figsize=(8,6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.coolwarm, edgecolors='k')
plt.title('Dataset')
plt.show()

print("A linear SVM will fail here because it is impossible to separate the two classes with a straight line.")

#Train linear classifier
svm_linear = SVC(kernel='linear', random_state=1).fit(X,y)
acc_linear = accuracy_score(y, svm_linear.predict(X))

#Plot decision boundary
fig,axes = plt.subplots(2,2,figsize=(12,6))
plot_decision_boundary(svm_linear,X,y,axes[0,1],"Linear SVM")
print(f"Accuracy of linear SVM: {acc_linear:.2f}")

#Train non-linear classifiers
svm_rbf_1 = SVC(kernel='rbf', gamma=1.0, random_state=1).fit(X,y)
acc_rbf_1 = accuracy_score(y, svm_rbf_1.predict(X))

svm_rbf_01 = SVC(kernel='rbf', gamma=0.1, random_state=1).fit(X,y)
acc_rbf_01 = accuracy_score(y, svm_rbf_01.predict(X))

svm_rbf_10 = SVC(kernel='rbf', gamma=10.0, random_state=1).fit(X,y)
acc_rbf_10 = accuracy_score(y, svm_rbf_10.predict(X))

#Plot decision boundaries
plot_decision_boundary(svm_rbf_1,X,y,axes[0,0], "RBF with gamma = 1.0")
plot_decision_boundary(svm_rbf_01,X,y,axes[0,1], "RBF with gamma = 0.1")
plot_decision_boundary(svm_rbf_10,X,y,axes[1,0], "RBF with gamma = 10.0")
print(f"Accuracy of RBF with gamma = 1.0: {acc_rbf_1:.2f}")
print("The RBF kernel solves the issue by using the kernel trick, which basically lifts the inner points onto a higher space so that they can be separated from the outer points.")
print("When gamma is too high, overfitting can occur because the model contours around individual training points.")
```

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import onnxruntime as ort

class TorchNet(nn.Module):
  def __init__(self, input_dim: int, output_dim: int=1):
    #Input layer: input_dim features
    #Hidden layer 1: 4 neurons
    #Hidden layer 2: 3 neurons
    #Output layer: output_dim neurons

    super().__init__()
    self.input_dim = input_dim
    self.output_dim = output_dim
    self.network = nn.Sequential(
        nn.Linear(self.input_dim, 4),
        nn.ReLU(),
        nn.Linear(4, 3),
        nn.ReLU(),
        nn.Linear(3, self.output_dim)
    )
    self.scaler = StandardScaler()
    self.loss_history = []

  def forward(self, x:torch.Tensor) -> torch.Tensor:
    return self.network(x)

  def load_and_standardize(self, csv_source: str):
    df = pd.read_csv(csv_source)
    X = df.drop(columns=['Outcome']).values
    y = df['Outcome'].values

    if X.shape[1] != self.input_dim:
      raise ValueError(f"Expected {self.input_dim} input features, but dataset has {X.shape[1]}.")

    X_scaled = self.scaler.fit_transform(X)
    X_tensor = torch.tensor(X_scaled, dtype=torch.float32)

    if self.output_dim == 1:
      y_tensor = torch.tensor(y, dtype=torch.float32).unsqueeze(1)
    else:
      y_tensor = torch.tensor(y, dtype=torch.long)

    return X_tensor, y_tensor

  def train_model(self, X_train, y_train, epochs=1000, lr=0.01):

    criterion = nn.BCEWithLogitsLoss() if self.output_dim == 1 else nn.CrossEntropyLoss()
    optimizer = optim.Adam(self.network.parameters(), lr=lr)
    self.train()
    self.loss_history = []

    for epoch in range(epochs):
      optimizer.zero_grad()
      outputs = self.forward(X_train)
      loss = criterion(outputs, y_train)
      loss.backward()
      optimizer.step()

      self.loss_history.append(loss.item())
      if (epoch+1) % 200 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")

  def plot_loss(self):
    plt.figure(figsize=(8,5))
    plt.plot(range(1,len(self.loss_history)+1), self.loss_history)
    plt.title("Training Loss vs Epochs")
    plt.xlabel("Epochs")
    plt.ylabel("Loss")
    plt.show()

  def save_onnx(self, path="diabetes_torchnet.onnx"):
    self.eval()
    dummy_input = torch.randn(1, self.input_dim)
    torch.onnx.export(
        self,
        dummy_input,
        path,
        export_params=True,
        opset_version=14,
        do_constant_folding=True,
        input_names=["input"],
        output_names=["output"],
        dynamic_axes={
            "input": {0: "batch_size"},
            "output": {0: "batch_size"}
        }
    )
    print(f"Model saved to {path}")

  def predict_from_onnx(self, path, sample_data) -> int:

    if len(sample_data) != self.input_dim:
      raise ValueError(f"Expected {self.input_dim} input features, but received {len(sample_data)}.")

    sample_array = np.array(sample_data, dtype=np.float32).reshape(1, -1)
    sample_scaled = self.scaler.transform(sample_array).astype(np.float32)

    session = ort.InferenceSession(path)
    input_name = session.get_inputs()[0].name
    logits = session.run(None, {input_name: sample_scaled})[0]

    if self.output_dim == 1:
      probability = 1.0/(1.0 + np.exp(-logits[0][0]))
      prediction = int(probability >= 0.5)
      print(f"Predicted Probability of Diabetes: {probability:.4f}")
    else:
      exp_logits = np.exp(logits)
      probabilities = exp_logits / exp_logits.sum(axis=1, keepdims=True)
      prediction = int(np.argmax(probabilities))
      print(f"Predicted Class Probabilities: {probabilities}")

    print(f"Predicted Outcome: {prediction}")
    return prediction

if __name__ == "__main__":
  DATA_URL = "https://raw.githubusercontent.com/rahulbhadani/Datasets/refs/heads/master/Diabetes/diabetes.csv"
  ONNX_PATH = "diabetes_torchnet.onnx"

  model = TorchNet(input_dim=8, output_dim=1)

  X_tensor, y_tensor = model.load_and_standardize(DATA_URL)
  model.train_model(X_tensor, y_tensor, epochs=1000, lr=0.01)
  model.plot_loss()
  model.save_onnx(ONNX_PATH)

  test_sample = [7, 149, 73, 94, 94, 32, 0.672, 45]
  predicted_class = model.predict_from_onnx(ONNX_PATH, test_sample)
```
